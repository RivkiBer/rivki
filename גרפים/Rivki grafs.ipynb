```python
time_study = [10,8,9,4,1,0,5,4,20,23,18,15,20,10,9,5,2,3,2,3]
grade = [100,98,97,95,76,67,50,78,82,85,100,100,99,99,85,89,92,59,66,47]
```

```python
plt.scatter(time_study, grade, alpha=0.6, cmap='magma')
plt.colorbar()  # הצגת סרגל הצבעים
plt.title('Scatter Plot Example')
plt.xlabel('time_study')
plt.ylabel('grade')
```
